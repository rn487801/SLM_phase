```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# Define Gaussian function
def gaussian(x, amp, cen, wid):
    return amp * np.exp(-(x - cen) ** 2 / (2 * wid ** 2))

# Define polynomial function (global)
def polynomial(x, *coeffs):
    return sum(c * x**i for i, c in enumerate(coeffs))

# Load data
file_path = 'C:/Users/rn487/OneDrive/Desktop/Importatnt Files/實驗室/Complex phonon mode/20240731_MoS2 Raman with OAM/RM60 AVG3_532LP_OAM0 to 5 and -1 to -5  Spectrum.dat'  # Replace with your file path
data = pd.read_csv(file_path, sep="\t", header=None)

# Extract Raman shift (first column) and intensity columns
raman_shift = data.iloc[:, 0].values
intensity_columns = data.iloc[:, 1:].values

# Define the fitting range (e.g., 350 to 450 cm^-1)
fitting_range = (300, 470)

# Polarization angles from 0 to 360 with 15 degree steps
angles = np.arange(0, 361, 15)

# Define initial guesses for the Gaussian parameters for E2g and A1g modes
initial_guesses = {
    'E2g': [1000, 368, 5],  # Example: [amplitude, center, width]
    'A1g': [1000, 387, 5]
}

# Define initial guesses for the background parameters
background_initial_guesses = [700, 0]  # Adjust the length for the polynomial order

# Store the fitted parameters for each angle
fitted_params = []

# Function to clean data by removing NaNs and infs
def clean_data(x, y):
    finite_mask = np.isfinite(y)
    return x[finite_mask], y[finite_mask]

# Function to filter data within the fitting range
def filter_range(x, y, range_min, range_max):
    range_mask = (x >= range_min) & (x <= range_max)
    return x[range_mask], y[range_mask]

# Define combined function to fit both background and peaks
def combined_function(x, *params):
    # Extract background and peak parameters
    background_params = params[:len(background_initial_guesses)]
    peak_params = params[len(background_initial_guesses):]
    
    # Evaluate polynomial background and Gaussian peaks
    background = polynomial(x, *background_params)
    peak1 = gaussian(x, *peak_params[:3])  # Parameters for the first peak (e.g., E2g)
    peak2 = gaussian(x, *peak_params[3:])  # Parameters for the second peak (e.g., A1g)
    
    return background + peak1 + peak2

def plot_and_fit(angle, intensity, initial_guesses, background_initial_guesses, bounds, output):
    print(f"Plot and fit for angle: {angle}")
    # Clean data
    raman_shift_clean, intensity_clean = clean_data(raman_shift, intensity)
    
    # Filter data within the fitting range
    raman_shift_filtered, intensity_filtered = filter_range(raman_shift_clean, intensity_clean, *fitting_range)
    
    # Check if the filtered data is not empty
    if len(intensity_filtered) == 0:
        print(f'Skipping angle {angle} degrees due to empty filtered data')
        return None
    
    # Fit combined background and peaks function
    try:
        popt_combined, _ = curve_fit(combined_function, raman_shift_filtered, intensity_filtered, 
                                      p0=background_initial_guesses + initial_guesses['E2g'] + initial_guesses['A1g'],
                                      bounds=bounds,
                                      maxfev=5000)
    except RuntimeError as e:
        print(f"Fit failed for angle {angle} degrees: {e}")
        popt_combined = None
    
    # Update the plot in the output widget
    with output:
        clear_output(wait=True)
        plt.figure(figsize=(10, 6))
        plt.plot(raman_shift_filtered, intensity_filtered, 'b-', label='Original Data')
        
        if popt_combined is not None:
            print(f"Fitting successful for angle: {angle}")
            background_params = popt_combined[:len(background_initial_guesses)]
            peak_params_E2g = popt_combined[len(background_initial_guesses):len(background_initial_guesses)+3]
            peak_params_A1g = popt_combined[len(background_initial_guesses)+3:]
            
            # Calculate background-subtracted curve
            background_curve = polynomial(raman_shift_filtered, *background_params)
            intensity_subtracted = intensity_filtered - background_curve
            
            plt.plot(raman_shift_filtered, background_curve, 'k--', label='Background Fit')
            plt.plot(raman_shift_filtered, gaussian(raman_shift_filtered, *peak_params_E2g), 'r--', label='E2g Fit')
            plt.plot(raman_shift_filtered, gaussian(raman_shift_filtered, *peak_params_A1g), 'g--', label='A1g Fit')
            plt.plot(raman_shift_filtered, combined_function(raman_shift_filtered, *popt_combined), 'm-', label='Combined Fit')
        else:
            print(f"Fitting failed for angle: {angle}")
        
        plt.title(f'Raman Spectra at {angle} Degrees')
        plt.xlabel('Raman Shift (cm^-1)')
        plt.ylabel('Intensity')
        plt.xlim(fitting_range)  # Set the x-axis limits to match the fitting range
        plt.legend()
        plt.show()
    
    return popt_combined

def create_table_input(initial_guesses, background_initial_guesses):
    print("Creating input table")
    # Create input widgets for background guesses
    background_inputs = [widgets.FloatText(value=val, description=f'b{i}', step=0.1) for i, val in enumerate(background_initial_guesses)]
    
    # Create input widgets for E2g guesses
    E2g_inputs = [widgets.FloatText(value=val, description=f'E2g_{param}', step=0.1) for param, val in zip(['amp', 'cen', 'wid'], initial_guesses['E2g'])]
    
    # Create input widgets for A1g guesses
    A1g_inputs = [widgets.FloatText(value=val, description=f'A1g_{param}', step=0.1) for param, val in zip(['amp', 'cen', 'wid'], initial_guesses['A1g'])]
    
    # Create checkboxes for locking parameters
    background_locks = [widgets.Checkbox(value=False, description=f'Lock b{i}') for i in range(len(background_initial_guesses))]
    E2g_locks = [widgets.Checkbox(value=False, description=f'Lock E2g_{param}') for param in ['amp', 'cen', 'wid']]
    A1g_locks = [widgets.Checkbox(value=False, description=f'Lock A1g_{param}') for param in ['amp', 'cen', 'wid']]
    
    return background_inputs, E2g_inputs, A1g_inputs, background_locks, E2g_locks, A1g_locks

def display_table_and_fit(angle, intensity):
    print(f"Displaying table and fitting for angle: {angle}")
    # Create input tables for background, E2g, and A1g
    background_inputs, E2g_inputs, A1g_inputs, background_locks, E2g_locks, A1g_locks = create_table_input(initial_guesses, background_initial_guesses)
    
    # Define bounds (initially allowing all parameters to vary)
    bounds_lower = [-np.inf] * (len(background_initial_guesses) + len(initial_guesses['E2g']) + len(initial_guesses['A1g']))
    bounds_upper = [np.inf] * (len(background_initial_guesses) + len(initial_guesses['E2g']) + len(initial_guesses['A1g']))
    
    output = widgets.Output()
    
    def update_fit(_):
        with output:
            print("Button clicked.")
            print(f"Updating fit for angle: {angle}")
            # Get current input values
            new_background_guesses = [input_widget.value for input_widget in background_inputs]
            print(new_background_guesses)
            new_initial_guesses_E2g = [input_widget.value for input_widget in E2g_inputs]
            new_initial_guesses_A1g = [input_widget.value for input_widget in A1g_inputs]
            
            new_initial_guesses = {
                'E2g': new_initial_guesses_E2g,
                'A1g': new_initial_guesses_A1g
            }
            
            # Update bounds based on locked parameters
            print(f"Bounds before update: {bounds_lower} to {bounds_upper}")
            epsilon = 1e-9  # Small value to ensure bounds are not equal
            
            for i, lock in enumerate(background_locks):
                print(f"Background lock {i}: {lock.value}")
                if lock.value:
                    print("background is locked")
                    bounds_lower[i] = new_background_guesses[i] - epsilon
                    bounds_upper[i] = new_background_guesses[i] + epsilon
            for i, lock in enumerate(E2g_locks):
                print(f"E2g lock {i}: {lock.value}")
                if lock.value:
                    print(new_initial_guesses_E2g)
                    bounds_lower[len(background_initial_guesses) + i] = new_initial_guesses_E2g[i] - epsilon
                    bounds_upper[len(background_initial_guesses) + i] = new_initial_guesses_E2g[i] + epsilon
            for i, lock in enumerate(A1g_locks):
                print(f"A1g lock {i}: {lock.value}")
                if lock.value:
                    print(new_initial_guesses_A1g)
                    bounds_lower[len(background_initial_guesses) + len(initial_guesses['E2g']) + i] = new_initial_guesses_A1g[i] - epsilon
                    bounds_upper[len(background_initial_guesses) + len(initial_guesses['E2g']) + i] = new_initial_guesses_A1g[i] + epsilon
            
            print(f"Bounds after update: {bounds_lower} to {bounds_upper}")
            
            # Fit with the new initial guesses and bounds
            popt_combined = plot_and_fit(angle, intensity, new_initial_guesses, new_background_guesses, (bounds_lower, bounds_upper), output)
            
            if popt_combined is not None:
                # Update the input fields with the new fitting results
                print(f"Updating input fields with new fitting results for angle: {angle}")
                for widget, val in zip(background_inputs, popt_combined[:len(background_initial_guesses)]):
                    widget.value = val
                for widget, val in zip(E2g_inputs, popt_combined[len(background_initial_guesses):len(background_initial_guesses)+3]):
                    widget.value = val
                for widget, val in zip(A1g_inputs, popt_combined[len(background_initial_guesses)+3:]):
                    widget.value = val
                
                # Store the fitting results
                fitted_params.append([angle] + list(popt_combined[len(background_initial_guesses):len(background_initial_guesses)+6]))
    
    fit_button = widgets.Button(description="Fit")
    fit_button.on_click(update_fit)

    
    # Display the input tables and fit button
    input_table = widgets.VBox([
        widgets.HBox([widgets.VBox([background_inputs[i], background_locks[i]]) for i in range(len(background_inputs))]),
        widgets.HBox([widgets.VBox([E2g_inputs[i], E2g_locks[i]]) for i in range(len(E2g_inputs))]),
        widgets.HBox([widgets.VBox([A1g_inputs[i], A1g_locks[i]]) for i in range(len(A1g_inputs))]),
        fit_button
    ])
    display(input_table)
    display(output)
    
    # Perform an initial fit
    update_fit(None)

summary_output = widgets.Output()

def summarize_results(_):
    with summary_output:
        print("Summarizing results")
        clear_output(wait=True)
        # Check if there are any fitting results to summarize
        if len(fitted_params) == 0:
            print("No fitting results to summarize.")
            return
        
        # Extract column names
        columns = ["Angle", "E2g_amp", "E2g_cen", "E2g_wid", "A1g_amp", "A1g_cen", "A1g_wid"]
        
        # Create a DataFrame from the fitted_params list
        summary_df = pd.DataFrame(fitted_params, columns=columns)
        
        # Replace existing rows for the same angle with the latest results
        summary_df = summary_df.groupby('Angle').tail(1)
        
        # Sort the DataFrame by the "Angle" column
        summary_df = summary_df.sort_values(by='Angle')
        
        # Display the summary DataFrame
        display(summary_df)

summarize_button = widgets.Button(description="Summarize Results")
summarize_button.on_click(summarize_results)



# Loop over each polarization angle and display the fitting interface
for i, angle in enumerate(angles):
    intensity = intensity_columns[:, i]
    
    display_table_and_fit(angle, intensity)

# Display the summarize button and output area
display(summarize_button)
display(summary_output)
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# Define Gaussian function
def gaussian(x, amp, cen, wid):
    return amp * np.exp(-(x - cen) ** 2 / (2 * wid ** 2))

# Define polynomial function (global)
def polynomial(x, *coeffs):
    return sum(c * x**i for i, c in enumerate(coeffs))

# Load data
file_path = 'F:/Optical skyrmion/20241023_optical skyrmion PL RM on encapMoS2 and Sapphire/encapsulated MoS2 PL5_grayscale70_l-5to5_delete4Spectrum.csv'  # Replace with your file path
data = pd.read_csv(file_path, sep=",", header=None)

# Extract wavelength (first column) and intensity columns
wavelength = data.iloc[:, 0].values
intensity_columns = data.iloc[:, 1:].values

# Define the fitting range
fitting_range = (1.6, 2.1)

# Example of other parameter instead of angles
other_parameter = np.arange(0, 11, 1)  # Replace with your parameter values

# Define initial guesses for the Gaussian parameters for A, B, and T peaks
initial_guesses = {
    'A': [1471, 1.8, 0.04],  # Example: [amplitude, center, width]
    'B': [779, 1.9, 0.08],
    'T': [1068, 1.78, 0.08]  # Example for the new peak
}

# Define initial guesses for the background parameters
background_initial_guesses = [611, 0]  # Adjust the length for the polynomial order

# Store the fitted parameters for each angle
fitted_params = []

# Function to clean data by removing NaNs and infs
def clean_data(x, y):
    finite_mask = np.isfinite(y)
    y = pd.to_numeric(y, errors='coerce')  # Converts non-numeric to NaN
    return x[finite_mask], y[finite_mask]

# Function to filter data within the fitting range
def filter_range(x, y, range_min, range_max):
    range_mask = (x >= range_min) & (x <= range_max)
    return x[range_mask], y[range_mask]

# Define combined function to fit both background and peaks
def combined_function(x, *params):
    # Extract background and peak parameters
    background_params = params[:len(background_initial_guesses)]
    peak_params = params[len(background_initial_guesses):]
    
    # Evaluate polynomial background and Gaussian peaks
    background = polynomial(x, *background_params)
    peak1 = gaussian(x, *peak_params[:3])  # Parameters for the first peak (e.g., A)
    peak2 = gaussian(x, *peak_params[3:6])  # Parameters for the second peak (e.g., B)
    peak3 = gaussian(x, *peak_params[6:])  # Parameters for the third peak (e.g., T)
    
    return background + peak1 + peak2 + peak3

def plot_and_fit(parameter, intensity, initial_guesses, background_initial_guesses, bounds, output):
    print(f"Plot and fit for parameter: {parameter}")
    # Clean data
    wavelength_clean, intensity_clean = clean_data(wavelength, intensity)
    
    # Filter data within the fitting range
    wavelength_filtered, intensity_filtered = filter_range(wavelength_clean, intensity_clean, *fitting_range)
    
    # Check if the filtered data is not empty
    if len(intensity_filtered) == 0:
        print(f'Skipping parameter {parameter} due to empty filtered data')
        return None
    
    # Fit combined background and peaks function
    try:
        popt_combined, _ = curve_fit(combined_function, wavelength_filtered, intensity_filtered, 
                                      p0=background_initial_guesses + initial_guesses['A'] + initial_guesses['B'] + initial_guesses['T'],
                                      bounds=bounds,
                                      maxfev=5000)
    except RuntimeError as e:
        print(f"Fit failed for parameter {parameter}: {e}")
        popt_combined = None
    
    # Update the plot in the output widget
    with output:
        clear_output(wait=True)
        plt.figure(figsize=(10, 6))
        plt.plot(wavelength_filtered, intensity_filtered, 'b-', label='Original Data')
        
        if popt_combined is not None:
            print(f"Fitting successful for parameter: {parameter}")
            background_params = popt_combined[:len(background_initial_guesses)]
            peak_params_A = popt_combined[len(background_initial_guesses):len(background_initial_guesses)+3]
            peak_params_B = popt_combined[len(background_initial_guesses)+3:len(background_initial_guesses)+6]
            peak_params_T = popt_combined[len(background_initial_guesses)+6:]
            
            # Calculate background-subtracted curve
            background_curve = polynomial(wavelength_filtered, *background_params)
            intensity_subtracted = intensity_filtered - background_curve
            
            plt.plot(wavelength_filtered, background_curve, 'k--', label='Background Fit')
            plt.plot(wavelength_filtered, gaussian(wavelength_filtered, *peak_params_A), 'r--', label='A Fit')
            plt.plot(wavelength_filtered, gaussian(wavelength_filtered, *peak_params_B), 'g--', label='B Fit')
            plt.plot(wavelength_filtered, gaussian(wavelength_filtered, *peak_params_T), 'b--', label='T Fit')
            plt.plot(wavelength_filtered, combined_function(wavelength_filtered, *popt_combined), 'm-', label='Combined Fit')
        else:
            print(f"Fitting failed for parameter: {parameter}")
        
        plt.title(f'PL Spectra at parameter: {parameter}')
        plt.xlabel('Wavelength (nm)')
        plt.ylabel('Intensity')
        plt.xlim(fitting_range)  # Set the x-axis limits to match the fitting range
        plt.legend()
        plt.show()
    
    return popt_combined

def create_table_input(initial_guesses, background_initial_guesses):
    print("Creating input table")
    # Create input widgets for background guesses
    background_inputs = [widgets.FloatText(value=val, description=f'b{i}', step=0.1) for i, val in enumerate(background_initial_guesses)]
    
    # Create input widgets for A guesses
    A_inputs = [widgets.FloatText(value=val, description=f'A_{param}', step=0.1) for param, val in zip(['amp', 'cen', 'wid'], initial_guesses['A'])]
    
    # Create input widgets for A1g guesses
    B_inputs = [widgets.FloatText(value=val, description=f'B_{param}', step=0.1) for param, val in zip(['amp', 'cen', 'wid'], initial_guesses['B'])]
    
    # Create input widgets for the new peak guesses
    T_inputs = [widgets.FloatText(value=val, description=f'T_{param}', step=0.1) for param, val in zip(['amp', 'cen', 'wid'], initial_guesses['T'])]
    
    # Create checkboxes for locking parameters
    background_locks = [widgets.Checkbox(value=False, description=f'Lock b{i}') for i in range(len(background_initial_guesses))]
    A_locks = [widgets.Checkbox(value=False, description=f'Lock A_{param}') for param in ['amp', 'cen', 'wid']]
    B_locks = [widgets.Checkbox(value=False, description=f'Lock B_{param}') for param in ['amp', 'cen', 'wid']]
    T_locks = [widgets.Checkbox(value=False, description=f'Lock T_{param}') for param in ['amp', 'cen', 'wid']]
    
    return background_inputs, A_inputs, B_inputs, T_inputs, background_locks, A_locks, B_locks, T_locks

def display_table_and_fit(parameter, intensity):
    print(f"Displaying table and fitting for parameter: {parameter}")
    # Create input tables for background, A, B, and T
    background_inputs, A_inputs, B_inputs, T_inputs, background_locks, A_locks, B_locks, T_locks = create_table_input(initial_guesses, background_initial_guesses)
    
    # Define bounds (initially allowing all parameters to vary)
    bounds_lower = [-np.inf] * (len(background_initial_guesses) + len(initial_guesses['A']) + len(initial_guesses['B']) + len(initial_guesses['T']))
    bounds_upper = [np.inf] * (len(background_initial_guesses) + len(initial_guesses['A']) + len(initial_guesses['B']) + len(initial_guesses['T']))
    
    output = widgets.Output()
    
    def update_fit(_):
        with output:
            print("Button clicked.")
            print(f"Updating fit for parameter: {parameter}")
            # Get current input values
            new_background_guesses = [input_widget.value for input_widget in background_inputs]
            print(new_background_guesses)
            new_initial_guesses_A = [input_widget.value for input_widget in A_inputs]
            new_initial_guesses_B = [input_widget.value for input_widget in B_inputs]
            new_initial_guesses_T = [input_widget.value for input_widget in T_inputs]
            
            new_initial_guesses = {
                'A': new_initial_guesses_A,
                'B': new_initial_guesses_B,
                'T': new_initial_guesses_T
            }
            
            # Update bounds based on locked parameters
            print(f"Bounds before update: {bounds_lower} to {bounds_upper}")
            epsilon = 1e-9  # Small value to ensure bounds are not equal
            
            for i, lock in enumerate(background_locks):
                print(f"Background lock {i}: {lock.value}")
                if lock.value:
                    print("background is locked")
                    bounds_lower[i] = new_background_guesses[i] - epsilon
                    bounds_upper[i] = new_background_guesses[i] + epsilon
            for i, lock in enumerate(A_locks):
                print(f"A lock {i}: {lock.value}")
                if lock.value:
                    print(new_initial_guesses_A)
                    bounds_lower[len(background_initial_guesses) + i] = new_initial_guesses_A[i] - epsilon
                    bounds_upper[len(background_initial_guesses) + i] = new_initial_guesses_A[i] + epsilon
            for i, lock in enumerate(B_locks):
                print(f"B lock {i}: {lock.value}")
                if lock.value:
                    print(new_initial_guesses_B)
                    bounds_lower[len(background_initial_guesses) + len(initial_guesses['A']) + i] = new_initial_guesses_B[i] - epsilon
                    bounds_upper[len(background_initial_guesses) + len(initial_guesses['A']) + i] = new_initial_guesses_B[i] + epsilon
            for i, lock in enumerate(T_locks):
                print(f"T lock {i}: {lock.value}")
                if lock.value:
                    print(new_initial_guesses_T)
                    bounds_lower[len(background_initial_guesses) + len(initial_guesses['A']) + len(initial_guesses['B']) + i] = new_initial_guesses_T[i] - epsilon
                    bounds_upper[len(background_initial_guesses) + len(initial_guesses['A']) + len(initial_guesses['B']) + i] = new_initial_guesses_T[i] + epsilon
            
            print(f"Bounds after update: {bounds_lower} to {bounds_upper}")
            
            # Fit with the new initial guesses and bounds
            popt_combined = plot_and_fit(parameter, intensity, new_initial_guesses, new_background_guesses, (bounds_lower, bounds_upper), output)
            
            if popt_combined is not None:
                # Update the input fields with the new fitting results
                print(f"Updating input fields with new fitting results for parameter: {parameter}")
                for widget, val in zip(background_inputs, popt_combined[:len(background_initial_guesses)]):
                    widget.value = val
                for widget, val in zip(A_inputs, popt_combined[len(background_initial_guesses):len(background_initial_guesses)+3]):
                    widget.value = val
                for widget, val in zip(B_inputs, popt_combined[len(background_initial_guesses)+3:len(background_initial_guesses)+6]):
                    widget.value = val
                for widget, val in zip(T_inputs, popt_combined[len(background_initial_guesses)+6:]):
                    widget.value = val
                
                # Store the fitting results
                fitted_params.append([parameter] + list(popt_combined[len(background_initial_guesses):len(background_initial_guesses)+9]))
    
    fit_button = widgets.Button(description="Fit")
    fit_button.on_click(update_fit)

    
    # Display the input tables and fit button
    input_table = widgets.VBox([
        widgets.HBox([widgets.VBox([background_inputs[i], background_locks[i]]) for i in range(len(background_inputs))]),
        widgets.HBox([widgets.VBox([A_inputs[i], A_locks[i]]) for i in range(len(A_inputs))]),
        widgets.HBox([widgets.VBox([B_inputs[i], B_locks[i]]) for i in range(len(B_inputs))]),
        widgets.HBox([widgets.VBox([T_inputs[i], T_locks[i]]) for i in range(len(T_inputs))]),
        fit_button
    ])
    display(input_table)
    display(output)
    
    # Perform an initial fit
    update_fit(None)

summary_output = widgets.Output()

def summarize_results(_):
    with summary_output:
        print("Summarizing results")
        clear_output(wait=True)
        # Check if there are any fitting results to summarize
        if len(fitted_params) == 0:
            print("No fitting results to summarize.")
            return
        
        # Extract column names
        columns = ["Parameter", "A_amp", "A_cen", "A_wid", "B_amp", "B_cen", "B_wid", "T_amp", "T_cen", "T_wid"]
        
        # Create a DataFrame from the fitted_params list
        summary_df = pd.DataFrame(fitted_params, columns=columns)
        
        # Replace existing rows for the same parameter with the latest results
        summary_df = summary_df.groupby('Parameter').tail(1)
        
        # Sort the DataFrame by the "Parameter" column
        summary_df = summary_df.sort_values(by='Parameter')
        
        # Display the summary DataFrame
        display(summary_df)

summarize_button = widgets.Button(description="Summarize Results")
summarize_button.on_click(summarize_results)

# Loop over each parameter and display the fitting interface
for i, parameter in enumerate(other_parameter):
    intensity = intensity_columns[:, i]
    
    display_table_and_fit(parameter, intensity)

# Display the summarize button and output area
display(summarize_button)
display(summary_output)
```
