```python
import cv2

# Load grayscale phase image
path = r"C:\Users\rn487\OneDrive\Desktop\grating L=0_.JPG"
img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image could not be loaded. Check the file path.")

# Create fullscreen window on SLM (second monitor)
cv2.namedWindow("SLM", cv2.WND_PROP_FULLSCREEN)
cv2.moveWindow("SLM", 1920, 0)  # adjust offset depending on your main screen width
cv2.setWindowProperty("SLM", cv2.WND_PROP_FULLSCREEN, cv2.WINDOW_FULLSCREEN)

# Show image
cv2.imshow("SLM", img)
```

```python
cv2.destroyAllWindows()
```

```python
import matplotlib.pyplot as plt
plt.imshow(img, cmap='gray')
plt.show()
```

```python
img_resized = cv2.resize(img, (1920, 1080), interpolation=cv2.INTER_NEAREST)
cv2.imshow("SLM", img_resized)
```
