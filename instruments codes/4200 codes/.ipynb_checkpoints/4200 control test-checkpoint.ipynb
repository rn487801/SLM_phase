```python
#import pyvisa
#import numpy
import time
from instrcomms import Communications
```

```python
#connect to 4200
INST_RESOURCE_STR = "TCPIP0::169.254.16.30::1225::SOCKET"
my4200 = Communications(INST_RESOURCE_STR)
```

```python
my4200.connect()  # opens connections to the 4200A-SCS
my4200._instrument_object.write_termination = "\0"  # Set PyVISA write terminator
my4200._instrument_object.read_termination = "\0"  # Set PyVISA read terminator
```

```python
#SMU1 current - SMU1 voltage @ SMU2 voltage

my4200.query("BC") # Clears all readings from the buffer
my4200.query("DE") # Channel definition page
my4200.query("CH1, 'VD', 'ID', 1, 1") 
my4200.query("CH2, 'VG', 'IG', 1, 2")
#my4200.query("CH3, 'VB', 'IB', 1, 2")
my4200.query("SS") # Source setup page
my4200.query("VR1, -5, 5, 0.1, 10e-6")
my4200.query("VP2, -10, 4, 10e-6")
#my4200.query("VC1, 0, 100e-3")
my4200.query("HT 0") # Hold time, delays the start of a sweep
my4200.query("DT 0.001") # Delay time, wait between the output voltage is set
my4200.query("IT2") # Integration time
my4200.query("RS 5") # Resolution in number of digits
my4200.query("RG 1, 1e-9") #Lowest range
my4200.query("RG 2, 1e-9")
#my4200.query("RG 3, 100e-9")
my4200.query("SM") # Measurment setup page
my4200.query("DM1") # Display modes
my4200.query("XN 'VD', 1, -5, 5") # X-axis
my4200.query("YA 'ID', 1, -1e-4, 1e-4") # Y-axis1
#my4200.query("YB 'ID', 1, -1e-4, 1e-4") # Y-axis2
my4200.query("MD") # Measurement control page
my4200.query("ME1") # Control measurment

# wait for measurement to complete
status = my4200.query("SP")

while int(status) != 1:
    status = my4200.query("SP")
    time.sleep(1)

# Save data
my4200.query("SV 'D FileName Comment'")
```

```python
my4200.disconnect() # close communications with the 4200A-SCS  
```

```python
#ID-VD@VG

my4200.connect() # opens connections to the 4200A-SCS
my4200.write("DE"+term) #Selects the channel definition page
# SMU, VoltageName, CurrentNamae, SourceMode(1.Volatge, 2.Current, 3.Common), SourceFunction(1.Sweep, 2.Step, 3.Constant, 4.VAR1' source function)
my4200.write("CH1, 'VD', 'ID', 1, 1"+term) 
my4200.write("CH2, 'VG', 'IG', 1, 2"+term)
#my4200.write("CH3, 'VG', 'IG', 1, 2"+term)
my4200.write("SS"+term) #Selects the source setup page
#VAR1 sweep
my4200.write("VR1, 0, 1, 5, 10e-6"+term) #VoltageSource(linear), Start, Step, StepNumber, Compliance
#VAR2 step
my4200.write("VP2, 5, 1, 10e-6"+term) #VoltageSource, Start, Stop, NumberOfStep, Compliance
#VAR3 constant
#my4200.write("VC3, 0, 100e-3"+term) #Source, OutputValue, Compliance
my4200.write("HT 0"+term) #Hold time
my4200.write("DT 0.001"+term) #Delay time, delay after voltage output
my4200.write("IT2"+term) #Integration time, 1: 0.1PLC, 2: 1PLC, 3: 10PLC, 4: Custom
my4200.write("RS 5"+term) #Resolution in digits
my4200.write("RG 1, 100e-9"+term) #Lowest range
my4200.write("RG 2, 100e-9"+term)
#my4200.write("RG 3, 100e-9"+term)
my4200.write("SM"+term) #Selects the measurement setup page
my4200.write("DM1"+term) #Display mode
my4200.write("XN 'VD', 1, 0, 5"+term) #Xaxis, scale(1.linear, 2.log), min, max
my4200.write("YA 'ID', 1, 0, 0.04"+term) #Yaxis
my4200.write("MD"+term) #Selects the measurement control page
my4200.write("ME1"+term) #Start measurement
status = my4200.query("SP"+term)
time.sleep(1)
#my4200.disconnect() # close communications with the 4200A-SCS
```
